```python
from pathlib import Path
import pandas as pd

csv_path = Path("../raw/styles.csv")

def fix_bad_line(fields):
    # Merge any overflow columns back into the last text column.
    if len(fields) > 10:
        return fields[:9] + [", ".join(fields[9:])]
    if len(fields) < 10:
        return fields + [""] * (10 - len(fields))
    return fields

df = pd.read_csv(
    csv_path,
    engine="python",
    on_bad_lines=fix_bad_line,
    encoding="utf-8",
)

print(f"Loaded {len(df):,} rows and {df.shape[1]} columns from {csv_path}")
df.head()
```
